## 8. 运行其余对照模型

下面的列表可以一次性跑完剩余模型。Colab 可能在一轮训练结束后断开，
重连时再次运行同一格即可跳过已完成运行。

```python
for model_name in [
    "unet_scratch",
    "resnet34_unet",
    "resnet34_transformer_decoder",
]:
    run_id = f"{model_name}_seed42"
    output = Path("outputs") / "runs" / run_id / "metrics.json"
    if output.exists():
        print("skip", run_id)
        continue
    subprocess.run([
        sys.executable,
        "-m",
        "seg_transformer.cli",
        "--config",
        "configs/base.json",
        "--override",
        "configs/colab_gpu.json",
        "train",
        "--model",
        model_name,
        "--seed",
        "42",
        "--run-id",
        run_id,
    ], check=True)
    subprocess.run([
        sys.executable,
        "-m",
        "seg_transformer.cli",
        "--config",
        "configs/base.json",
        "evaluate",
        "--run-id",
        run_id,
    ], check=True)
```

## 9. 补跑 seed 7 与 seed 123

单种子只能说明一次运行，不能估计初始化波动。本节会把四个模型分别用
seed 7 和 seed 123 再运行，已经完成的实验会自动跳过。

```python
for seed in [7, 123]:
    for model_name in [
        "unet_scratch",
        "resnet34_unet",
        "resnet34_transformer",
        "resnet34_transformer_decoder",
    ]:
        run_id = f"{model_name}_seed{seed}"
        output = Path("outputs") / "runs" / run_id / "metrics.json"
        if output.exists():
            print("skip", run_id)
            continue
        subprocess.run([
            sys.executable,
            "-m",
            "seg_transformer.cli",
            "--config",
            "configs/base.json",
            "--override",
            "configs/colab_gpu.json",
            "train",
            "--model",
            model_name,
            "--seed",
            str(seed),
            "--run-id",
            run_id,
        ], check=True)
        subprocess.run([
            sys.executable,
            "-m",
            "seg_transformer.cli",
            "--config",
            "configs/base.json",
            "evaluate",
            "--run-id",
            run_id,
        ], check=True)
```

## 12. 结果解释边界

- 只使用固定的 96 张原始图像子集，不等同于完整 DeepGlobe 基准。
- 按原图切分可以避免图块泄漏，但仍不是按地区或采集条带划分。
- 128x128 输入会损失细节，mIoU 不能直接与 1024x1024 原报告比较。
- Transformer 架构本身不是新贡献，重点是迁移学习、全局上下文、逐类混淆和计算成本的对照。
